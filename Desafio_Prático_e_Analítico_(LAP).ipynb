**a) Refaça o código utilizando a estrutura FOR**

```python
numeros = [0, 0, 0, 0, 0]

for x in range(5):
    numeros[x] = float(input("Digite um número qualquer: "))

for y in range(5):
    print(numeros[y])
```

```python
numeros = []

quantidade = int(input("Quantos números deseja inserir? "))

for i in range(quantidade):
    numero = float(input("Digite um número qualquer: "))
    numeros.append(numero)

for numero in numeros:
    print(numero)
```

**4. Faça um programa que percorra duas listas, A e B, e gere uma terceira sem
elementos repetidos.
A = [‘Python’, ‘Java’, ‘C’, ‘PHP’, ‘JavaScript’, ‘Dart’]
B = [‘C++’, ‘Python’, ‘Java’, ‘Julia’, ‘Go’, ‘JavaScript’]**

```python
a = ['Python', 'Java', 'C', 'PHP', 'JavaScript', 'Dart']
b = ['C++', 'Python', 'Java', 'Julia', 'Go', 'JavaScript']

c = []

for x in a:
    if x not in c:
        c.append(x)

for x in b:
    if x not in c:
        c.append(x)

print(c)
```

**5. Observe o código a seguir e analise as questões pedidas**

**a) Explique a função das linhas 20, 24, 25 e 27**

R= A linha 20 (achou=False) serve para criar uma variável que indica que o valor ainda não foi encontrado.

A linha 24 (achou=True) muda o valor para verdadeiro quando o número procurado é encontrado.

A linha 25 (break) para o while, porque o número já foi encontrado e não precisa continuar procurando.

A linha 27 (if achou:) verifica se o número foi encontrado para mostrar a mensagem correta.

**b) Refaça o código de forma a realizar a mesma tarefa, mas sem utilizar a
variável achou**

```python
L = [15, 7, 27, 39]
p = int(input("Digite o valor a procurar: "))

x = 0
while x < len(L):
    if L[x] == p:
        print(f"{p} achado na posição {x}")
        break
    x += 1
else:
    print(f"{p} não encontrado")
```

**6. Observe o código a seguir e analise as questões pedidas**

**a) Modifique o código usando a estrutura FOR para as duas repetições.**

```python
L = []

while True:
    n = int(input("Digite um número ou 0 para sair: "))

    if n == 0:
        break

    L.append(n)

for x in L:
    print(x)
```

**b) A alternativa anterior teve êxito? Justifique.**

R= Sim, a alternativa teve êxito, porque o for consegue percorrer todos os elementos da lista sem precisar controlar a posição usando uma variável como x.

**8. Escreva um código que receba números ou palavras. O código deve
armazenar cada situação em listas diferentes, ou seja, uma lista só para as
palavras e outra só para os números. O programa deve encerrar quando o
usuário digitar o x ou X. No fim, mostre o que foi inserido nas duas listas, a
quantidade total de coisas digitadas pelo usuário. (Dica: métodos em Python)**

```python
palavras = []
numeros = []

while True:
    valor = input("Digite uma palavra ou número (X para sair): ")

    if valor.upper() == "X":
        break

    if valor.isnumeric():
        numeros.append(valor)
    else:
        palavras.append(valor)

print("Palavras:", palavras)
print("Números:", numeros)
print("Total digitado:", len(palavras) + len(numeros))
```
